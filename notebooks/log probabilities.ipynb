```python
import numpy as np

from typing import Optional

Array = np.ndarray
```

```python
class GaussianNoise:
    """
    Gaussian noise observation model.
    y_i = f(x_i) + epsilon_i
    epsilon_i ~ N(0, sigma2)
    """

    def __init__(self, sigma2: float):
        self.sigma2: float = float(sigma2)

    @property
    def pref(self) -> float:
        return 1 / np.sqrt(2 * np.pi * self.sigma2)

    def log_prob(self, residual) -> float:
        """Fully normalized log-likelihood."""
        lprob: float = np.log(self.pref) * -((0.5 * self.sigma2) * residual**2).sum()
        return lprob

    def energy(self, residual) -> float:
        return -self.log_prob(residual)
```

```python
class GaussianNoiseOld:
    """
    Gaussian noise observation model.
    y_i = f(x_i) + epsilon_i
    epsilon_i ~ N(0, sigma2)
    """

    def __init__(self, sigma2: float):
        self.sigma2: float = float(sigma2)

    def log_prob(self, residual) -> float:
        """Fully normalized log-likelihood."""
        n = residual.size
        # log N(y; f(x), sigma2*I) = -n/2 log(2*pi*sigma2) - 1/(2*sigma2) sum (y - f(x))^2
        return -0.5 * n * np.log(2.0 * np.pi * self.sigma2) - (
            0.5 / self.sigma2
        ) * np.sum(residual**2)

    def energy(self, residual) -> float:
        """Paper's E(theta): 1/(2n) * sum squared residuals."""
        # residual = y - prediction
        # return float(0.5 * np.mean(residual**2))
        return -self.log_prob(residual) / residual.size
```

```python
gn = GaussianNoise(sigma2=0.1)
gn_old = GaussianNoiseOld(sigma2=0.1)
```

```python
print(gn.log_prob(np.arange(20)))
print(gn_old.log_prob(np.arange(20)))
```
